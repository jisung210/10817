```python
from gtts import gTTS

# 1. 변환할 텍스트와 언어 설정
text = "안녕하세요.gTTS를 이용해 텍스트를 음성으로 변환하고 있습니다."
tts = gTTS(text=text,lang='ko')

# 2. mp3 파일로 저장
tts.save("output.mp3")
print("음성 파일 저장 완료!")
```

```python
from gtts import gTTS
from IPython.display import Audio

while True:
# 1. 변환할 텍스트와 언어 설정
  text = input('텍스트를 입력하세요:')
  tts = gTTS(text=text,lang='ko')
  if text=='종료':
    break
# 2. mp3 파일로 저장
  tts.save("output1.mp3")


  Audio("output1.mp3",autoplay=True)
```
