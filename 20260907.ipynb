```python
from datetime import datetime, date, time
print('<디데이(D-day) 프로그램')
print()
year = int(input('연도를 입력하세요:'))
month = int(input('월을 입력하세요:'))
day = int(input('일을 입력하세요:'))
dt4 = date(year,month,day)
today = date.today()
dday = (dt4-today).days
print(f'{year}년 {month}월 {day}일은 {dday}일 남았습니다.')
```

```python
a = int(input('빨간색 접시의 개수를 입력하세요:'))
b = int(input('노란색 접시의 개수를 입력하세요:'))
c = int(input('파란색 접시의 개수를 입력하세요:'))
d = int(input('초록색 접시의 개수를 입력하세요:'))
a1 = (1000*a)
b1 = (2000*b)
c1 = (3000*c)
d1 = (4000*d)
all = (a1+b1+c1+d1)
print('====회전 초밥 영수증===')
print('접시   단가 개수 합계')
print(f'빨간색 1000 {a} {a1}')
print(f'노란색 2000 {b} {b1}')
print(f'파란색 3000 {c} {c1}')
print(f'초록색 4000 {d} {d1}')
print(f'총 가격       {all}')
print('=======================')
```
