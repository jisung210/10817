```python
a = input('첫 번째 영어 문장을 입력하세요:')
b = input('두 번째 영어 문장을 입력하세요:')
c = input('세 번째 영어 문장을 입력하세요:')
text = a + '' + b + '' +c
text = text.replace(',','').replace('.','')
text = text.lower()
words = text.split()
unique_words = set(words)
sorted_unique_words = sorted(unique_words)
print('\n 단어 목록')
print(sorted_unique_words)
```

```python
import random
a = ['김민준','이서준','박도윤','최시우','정예준','강주원','조유준']
b = int(input('뽑을 당첨자 수를 입력하세요:'))
winers = random.sample(a,b)
winers.sort
print(winers)
```
