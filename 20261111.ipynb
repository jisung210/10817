```python
def a (q,w,e):
  if w == '+':
    return q+e
  elif w == '-':
    return q-e
  elif w == '*':
    return q*e
  elif w == '/':
    if e == 0:
      return
    elif q / e:
      return q / e
  else :
    return



print(a(10,'+',5))
print(a(10,'-',5))
print(a(10,'*',5))
print(a(10,'/',5))
```
