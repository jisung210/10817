```python
def qwer(a):
  if a%2==0:
    return '짝수'
  elif a%2==1:
    return '홀수'

b = int(input('정수를 입력하세요'))
print(qwer(b))
```

```python
def abcd(w,e):
  q=(w * e)/2
  return q

b = int(input('밑변을 입력하세요'))
c = int(input('높이를 입력하세요'))
print(abcd(b,c))
```
