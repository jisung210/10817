```python
a = [1,2,3,10,20]
print('원본:',a)
a[2] = 4
print('결과1 :',a)
a[3:] = [5,6]             #2. 슬라이싱 수정
print('결과2 :', a)
a[2:] = [3,4]             #3. 슬라이싱 수정(요소 개수와 다름.)
print('결과3 :', a)
del a[1]
print('결과4 :', a)
del a[1:]
print('결과5 :', a)
del a                     #리스트 자체 삭제
```
